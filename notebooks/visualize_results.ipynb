```python
from pathlib import Path
import geopandas as gpd

# Dynamically locate the project root directory
BASE_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
GEOJSON_PATH = BASE_DIR / "outputs" / "vectors" / "parcel_risk_assessment.geojson"

print(f"Loading from: {GEOJSON_PATH}")

# Load the GeoJSON file
risk_gdf = gpd.read_file(GEOJSON_PATH)

print(f"Total structures assessed: {len(risk_gdf)}")
print(f"High risk structures (Risk Score > 5.0): {len(risk_gdf[risk_gdf['risk_score'] > 5.0])}")
risk_gdf.head()
```

```python
from pathlib import Path
import leafmap

BASE_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
RASTER_PATH = str(BASE_DIR / "data" / "raw" / "aerial_sample.tif")

# Initialize interactive map centered over San Jose, CA
m = leafmap.Map(center=[37.3337, -121.8855], zoom=18)

# Add satellite raster layer
m.add_raster(RASTER_PATH, layer_name="Aerial Image (Raw)")

# Overlay vectorized building risk polygons
m.add_gdf(
    risk_gdf,
    layer_name="Building Risk Assessment",
    style={"color": "#ff0000", "weight": 2, "fillOpacity": 0.3}
)

m
```
